```python
# ============================================================
# RAM-FRIENDLY STRONG EXTRA TREES PIPELINE
# Traffic + Weather + Time + Events
#
# IMPORTANT:
# This version DOES NOT create traffic_long.
# ============================================================

import gc
import numpy as np
import pandas as pd

from sklearn.preprocessing import LabelEncoder


# ============================================================
# 1. BASIC DATA
# ============================================================

df = merged

df['time'] = pd.to_datetime(
    df['time']
)

df = df.sort_values(
    'time'
).reset_index(
    drop=True
)


# ============================================================
# 2. SENSOR MATRIX
# ============================================================

sensor_columns = [
    col
    for col in df.columns
    if str(col).isdigit()
]

N_SENSORS = len(sensor_columns)
N_TIMESTAMPS = len(df)


print(
    'Number of sensors:',
    N_SENSORS
)

print(
    'Number of timestamps:',
    N_TIMESTAMPS
)


# ------------------------------------------------------------
# IMPORTANT:
# Convert ONLY sensor matrix to float32
# ------------------------------------------------------------

speed_matrix = (
    df[sensor_columns]
    .to_numpy(
        dtype=np.float32,
        copy=True
    )
)


print(
    'Speed matrix:',
    speed_matrix.shape
)

print(
    'Speed matrix RAM:',
    round(
        speed_matrix.nbytes / 1024**2,
        2
    ),
    'MB'
)


# ============================================================
# 3. TIME FEATURES
# ============================================================

times = df['time']

hour = (
    times.dt.hour
    .to_numpy(
        dtype=np.int8
    )
)

minute = (
    times.dt.minute
    .to_numpy(
        dtype=np.int8
    )
)

dow = (
    times.dt.dayofweek
    .to_numpy(
        dtype=np.int8
    )
)

month = (
    times.dt.month
    .to_numpy(
        dtype=np.int8
    )
)


is_weekend = (
    dow >= 5
).astype(
    np.int8
)


is_holiday = (
    df['Holiday']
    .ne('None')
    .to_numpy(
        dtype=np.int8
    )
)


# ============================================================
# 4. CYCLIC TIME
# ============================================================

hour_sin = np.sin(
    2 * np.pi * hour / 24
).astype(
    np.float32
)

hour_cos = np.cos(
    2 * np.pi * hour / 24
).astype(
    np.float32
)

dow_sin = np.sin(
    2 * np.pi * dow / 7
).astype(
    np.float32
)

dow_cos = np.cos(
    2 * np.pi * dow / 7
).astype(
    np.float32
)

month_sin = np.sin(
    2 * np.pi * month / 12
).astype(
    np.float32
)

month_cos = np.cos(
    2 * np.pi * month / 12
).astype(
    np.float32
)


# ============================================================
# 5. WEATHER
# ============================================================

temperature = (
    df['temperature_2m']
    .to_numpy(
        dtype=np.float32
    )
)

humidity = (
    df['relative_humidity_2m']
    .to_numpy(
        dtype=np.float32
    )
)

rain = (
    df['rain']
    .to_numpy(
        dtype=np.float32
    )
)

weather_code = (
    df['weather_code']
    .to_numpy(
        dtype=np.float32
    )
)


temp_humidity = (
    temperature
    *
    humidity
).astype(
    np.float32
)


rain_flag = (
    rain > 0
).astype(
    np.int8
)


rain_intensity = np.log1p(
    np.maximum(
        rain,
        0
    )
).astype(
    np.float32
)


bad_weather = (
    (rain > 0)
    |
    (weather_code >= 50)
).astype(
    np.int8
)


# ============================================================
# 6. RUSH HOUR
# ============================================================

morning_rush = (
    (hour >= 6)
    &
    (hour <= 9)
).astype(
    np.int8
)


evening_rush = (
    (hour >= 15)
    &
    (hour <= 19)
).astype(
    np.int8
)


rush_hour = (
    (morning_rush == 1)
    |
    (evening_rush == 1)
).astype(
    np.int8
)


# ============================================================
# 7. EVENT FEATURES
# ============================================================

events_data = events

event_start = pd.to_datetime(
    events_data[
        'Event Date & Time Start'
    ],
    errors='coerce'
)


event_start = (
    event_start
    .dropna()
)


# Count events per hour
event_hour_counts = (
    event_start
    .dt.floor('h')
    .value_counts()
)


timestamp_hours = (
    times.dt.floor('h')
)


event_count_time = (
    timestamp_hours
    .map(
        event_hour_counts
    )
    .fillna(0)
    .to_numpy(
        dtype=np.float32
    )
)


has_event_time = (
    event_count_time > 0
).astype(
    np.int8
)


print(
    'Valid events:',
    len(event_start)
)


# ============================================================
# 8. VALID TIME RANGE
# ============================================================

# Largest lag = 24
# Target = +3
#
# Therefore:
#
# t >= 24
# t <= N - 4

START = 24
END = N_TIMESTAMPS - 3

valid_times = np.arange(
    START,
    END
)

N_VALID = len(
    valid_times
)


print(
    'Valid timestamps:',
    N_VALID
)


# ============================================================
# 9. TARGET
# ============================================================

target_matrix = (
    speed_matrix[
        valid_times + 3
    ]
)


# ============================================================
# 10. SENSOR ENCODING
# ============================================================

sensor_ids = np.array(
    [
        int(x)
        for x in sensor_columns
    ],
    dtype=np.int32
)


# Since sensors already have a stable
# integer position, we don't need LabelEncoder.
#
# 0 ... 206 is enough.

sensor_encoded = np.arange(
    N_SENSORS,
    dtype=np.int16
)


# ============================================================
# 11. FEATURE COUNT
# ============================================================

feature_names = [

    'sensor_id_encoded',

    'speed',

    'lag_1',
    'lag_2',
    'lag_3',
    'lag_4',
    'lag_6',
    'lag_9',
    'lag_12',
    'lag_18',
    'lag_24',

    'rolling_mean_3',
    'rolling_mean_6',
    'rolling_mean_12',
    'rolling_mean_24',

    'rolling_std_6',
    'rolling_std_12',
    'rolling_std_24',

    'speed_change_1',
    'speed_change_3',
    'speed_change_6',

    'speed_vs_mean_6',
    'speed_vs_mean_12',

    'temperature_2m',
    'relative_humidity_2m',
    'rain',
    'weather_code',

    'temp_humidity_interaction',
    'rain_flag',
    'rain_intensity',
    'bad_weather',

    'hour',
    'minute',
    'day_of_week',
    'month',

    'is_weekend',
    'is_holiday',

    'is_morning_rush',
    'is_evening_rush',
    'is_rush_hour',

    'hour_sin',
    'hour_cos',
    'dow_sin',
    'dow_cos',
    'month_sin',
    'month_cos',

    'event_count',
    'has_event'
]


N_FEATURES = len(
    feature_names
)


print(
    'Number of features:',
    N_FEATURES
)


# ============================================================
# 12. TRAIN / TEST TIME SPLIT
# ============================================================

split_position = int(
    N_VALID * 0.8
)


train_time_positions = (
    valid_times[
        :split_position
    ]
)

test_time_positions = (
    valid_times[
        split_position:
    ]
)


print(
    'Train timestamps:',
    len(
        train_time_positions
    )
)

print(
    'Test timestamps:',
    len(
        test_time_positions
    )
)


# ============================================================
# 13. MEMORY-FRIENDLY FEATURE FUNCTION
# ============================================================

def build_features_for_times(
    time_positions
):

    n_t = len(
        time_positions
    )

    n_rows = (
        n_t
        *
        N_SENSORS
    )


    X = np.empty(
        (
            n_rows,
            N_FEATURES
        ),
        dtype=np.float32
    )


    # --------------------------------------------------------
    # row mapping
    # --------------------------------------------------------

    # time-major:
    #
    # t0 sensor0
    # t0 sensor1
    # ...
    # t1 sensor0
    # ...

    row = 0


    # --------------------------------------------------------
    # Sensor ID
    # --------------------------------------------------------

    X[:, 0] = np.tile(
        sensor_encoded,
        n_t
    )


    # --------------------------------------------------------
    # Current speed
    # --------------------------------------------------------

    X[:, 1] = speed_matrix[
        time_positions
    ].reshape(
        -1
    )


    # --------------------------------------------------------
    # Lags
    # --------------------------------------------------------

    lag_values = [
        1,
        2,
        3,
        4,
        6,
        9,
        12,
        18,
        24
    ]


    col = 2


    for lag in lag_values:

        X[:, col] = (
            speed_matrix[
                time_positions - lag
            ]
            .reshape(-1)
        )

        col += 1


    # --------------------------------------------------------
    # Rolling statistics
    # --------------------------------------------------------

    speed_df = pd.DataFrame(
        speed_matrix
    )


    shifted = (
        speed_df
        .shift(1)
    )


    rolling_windows = [
        3,
        6,
        12,
        24
    ]


    # Rolling means
    for window in rolling_windows:

        rolling_mean = (
            shifted
            .rolling(
                window,
                min_periods=window
            )
            .mean()
            .to_numpy(
                dtype=np.float32
            )
        )

        X[:, col] = (
            rolling_mean[
                time_positions
            ]
            .reshape(-1)
        )

        col += 1

        del rolling_mean


    # Rolling std
    for window in [
        6,
        12,
        24
    ]:

        rolling_std = (
            shifted
            .rolling(
                window,
                min_periods=window
            )
            .std()
            .to_numpy(
                dtype=np.float32
            )
        )

        X[:, col] = (
            rolling_std[
                time_positions
            ]
            .reshape(-1)
        )

        col += 1

        del rolling_std


    del shifted
    del speed_df

    gc.collect()


    # --------------------------------------------------------
    # Traffic dynamics
    # --------------------------------------------------------

    current = speed_matrix[
        time_positions
    ]

    lag1 = speed_matrix[
        time_positions - 1
    ]

    lag3 = speed_matrix[
        time_positions - 3
    ]

    lag6 = speed_matrix[
        time_positions - 6
    ]


    X[:, col] = (
        current - lag1
    ).reshape(-1)

    col += 1


    X[:, col] = (
        current - lag3
    ).reshape(-1)

    col += 1


    X[:, col] = (
        current - lag6
    ).reshape(-1)

    col += 1


    # --------------------------------------------------------
    # Relative speed
    # --------------------------------------------------------

    # Recompute only needed rolling means
    speed_df = pd.DataFrame(
        speed_matrix
    )

    shifted = speed_df.shift(1)


    rolling6 = (
        shifted
        .rolling(
            6,
            min_periods=6
        )
        .mean()
        .to_numpy(
            dtype=np.float32
        )
    )


    rolling12 = (
        shifted
        .rolling(
            12,
            min_periods=12
        )
        .mean()
        .to_numpy(
            dtype=np.float32
        )
    )


    X[:, col] = (
        current
        /
        (
            rolling6[
                time_positions
            ]
            + 1e-6
        )
    ).reshape(-1)

    col += 1


    X[:, col] = (
        current
        /
        (
            rolling12[
                time_positions
            ]
            + 1e-6
        )
    ).reshape(-1)

    col += 1


    del rolling6
    del rolling12
    del shifted
    del speed_df

    gc.collect()


    # --------------------------------------------------------
    # Weather
    # --------------------------------------------------------

    X[:, col] = np.repeat(
        temperature[
            time_positions
        ],
        N_SENSORS
    )

    col += 1


    X[:, col] = np.repeat(
        humidity[
            time_positions
        ],
        N_SENSORS
    )

    col += 1


    X[:, col] = np.repeat(
        rain[
            time_positions
        ],
        N_SENSORS
    )

    col += 1


    X[:, col] = np.repeat(
        weather_code[
            time_positions
        ],
        N_SENSORS
    )

    col += 1


    X[:, col] = np.repeat(
        temp_humidity[
            time_positions
        ],
        N_SENSORS
    )

    col += 1


    X[:, col] = np.repeat(
        rain_flag[
            time_positions
        ],
        N_SENSORS
    )

    col += 1


    X[:, col] = np.repeat(
        rain_intensity[
            time_positions
        ],
        N_SENSORS
    )

    col += 1


    X[:, col] = np.repeat(
        bad_weather[
            time_positions
        ],
        N_SENSORS
    )

    col += 1


    # --------------------------------------------------------
    # Time
    # --------------------------------------------------------

    time_arrays = [
        hour,
        minute,
        dow,
        month,
        is_weekend,
        is_holiday,
        morning_rush,
        evening_rush,
        rush_hour,
        hour_sin,
        hour_cos,
        dow_sin,
        dow_cos,
        month_sin,
        month_cos
    ]


    for arr in time_arrays:

        X[:, col] = np.repeat(
            arr[
                time_positions
            ],
            N_SENSORS
        )

        col += 1


    # --------------------------------------------------------
    # Events
    # --------------------------------------------------------

    X[:, col] = np.repeat(
        event_count_time[
            time_positions
        ],
        N_SENSORS
    )

    col += 1


    X[:, col] = np.repeat(
        has_event_time[
            time_positions
        ],
        N_SENSORS
    )

    col += 1


    print(
        'Features built:',
        X.shape
    )

    print(
        'RAM:',
        round(
            X.nbytes / 1024**2,
            2
        ),
        'MB'
    )


    return X


# ============================================================
# 14. TARGET FUNCTION
# ============================================================

def build_target(
    time_positions
):

    return (
        speed_matrix[
            time_positions + 3
        ]
        .reshape(-1)
        .astype(
            np.float32
        )
    )


# ============================================================
# 15. TRAIN SAMPLE ONLY
# ============================================================

TRAIN_SAMPLES = 300_000


rng = np.random.default_rng(
    42
)


total_train_rows = (
    len(train_time_positions)
    *
    N_SENSORS
)


sample_size = min(
    TRAIN_SAMPLES,
    total_train_rows
)


sample_indices = (
    rng.choice(
        total_train_rows,
        size=sample_size,
        replace=False
    )
)


# Convert flat indices to:
# time position + sensor

sample_time_idx = (
    sample_indices
    // N_SENSORS
)

sample_sensor_idx = (
    sample_indices
    % N_SENSORS
)


sample_times = (
    train_time_positions[
        sample_time_idx
    ]
)


print()
print(
    'Training samples:',
    sample_size
)


# ============================================================
# 16. BUILD TRAIN SAMPLE
# ============================================================

X_train_full = build_features_for_times(
    sample_times
)


# We built all sensors for each selected timestamp.
# Select only the exact random sensor rows.

row_indices = (
    sample_time_idx
    *
    N_SENSORS
    +
    sample_sensor_idx
)


X_train = (
    X_train_full[
        row_indices
    ]
)


y_train = (
    speed_matrix[
        sample_times + 3,
        sample_sensor_idx
    ]
    .astype(
        np.float32
    )
)


del X_train_full
del row_indices
del sample_indices
del sample_time_idx
del sample_sensor_idx
del sample_times

gc.collect()


print(
    'FINAL X_train:',
    X_train.shape
)

print(
    'FINAL y_train:',
    y_train.shape
)

print(
    'X_train RAM:',
    round(
        X_train.nbytes / 1024**2,
        2
    ),
    'MB'
)


# ============================================================
# 17. TEST DATA
# ============================================================

print()
print(
    'Building test features...'
)


X_test = build_features_for_times(
    test_time_positions
)


y_test = build_target(
    test_time_positions
)


print(
    'X_test:',
    X_test.shape
)

print(
    'y_test:',
    y_test.shape
)


# ============================================================
# 18. CLEAN UNUSED DATA
# ============================================================

del df

gc.collect()


# ============================================================
# 19. FINAL CHECK
# ============================================================

print()
print(
    '=' * 55
)

print(
    'RAM-FRIENDLY DATASET READY'
)

print(
    '=' * 55
)

print(
    'Features:',
    N_FEATURES
)

print(
    'X_train:',
    X_train.shape
)

print(
    'y_train:',
    y_train.shape
)

print(
    'X_test:',
    X_test.shape
)

print(
    'y_test:',
    y_test.shape
)

print(
    'X_train dtype:',
    X_train.dtype
)

print(
    'X_test dtype:',
    X_test.dtype
)

print(
    'X_train RAM:',
    round(
        X_train.nbytes / 1024**2,
        2
    ),
    'MB'
)

print(
    'X_test RAM:',
    round(
        X_test.nbytes / 1024**2,
        2
    ),
    'MB'
)

print(
    '=' * 55
)
```

```python
# ============================================================
# REBUILD TRAIN DATA - EXACT X/Y PAIRING
# ============================================================

import numpy as np
import gc


# ------------------------------------------------------------
# Settings
# ------------------------------------------------------------

TRAIN_SAMPLES = 300_000

rng = np.random.default_rng(42)


# ------------------------------------------------------------
# Random (time, sensor) pairs
# ------------------------------------------------------------

random_time_positions = rng.choice(
    train_time_positions,
    size=TRAIN_SAMPLES,
    replace=True
)

random_sensor_positions = rng.integers(
    0,
    N_SENSORS,
    size=TRAIN_SAMPLES
)


print(
    'Samples:',
    TRAIN_SAMPLES
)


# ============================================================
# BUILD X AND y FROM THE SAME INDICES
# ============================================================

t = random_time_positions
s = random_sensor_positions


# ------------------------------------------------------------
# Allocate
# ------------------------------------------------------------

X_train_new = np.empty(
    (
        TRAIN_SAMPLES,
        N_FEATURES
    ),
    dtype=np.float32
)


# ============================================================
# 1. SENSOR
# ============================================================

X_train_new[:, 0] = (
    sensor_encoded[s]
)


# ============================================================
# 2. CURRENT SPEED
# ============================================================

X_train_new[:, 1] = (
    speed_matrix[t, s]
)


# ============================================================
# 3. LAGS
# ============================================================

lag_list = [
    1,
    2,
    3,
    4,
    6,
    9,
    12,
    18,
    24
]


col = 2


for lag in lag_list:

    X_train_new[:, col] = (
        speed_matrix[
            t - lag,
            s
        ]
    )

    col += 1


# ============================================================
# 4. ROLLING FEATURES
# ============================================================

# We calculate these only for the requested
# (time, sensor) pairs instead of creating
# a gigantic long dataframe.


def rolling_mean_for_pairs(
    times,
    sensors,
    window
):

    result = np.empty(
        len(times),
        dtype=np.float32
    )

    for i in range(
        len(times)
    ):

        tt = times[i]
        ss = sensors[i]

        result[i] = np.mean(
            speed_matrix[
                tt - window:
                tt,
                ss
            ]
        )

    return result


def rolling_std_for_pairs(
    times,
    sensors,
    window
):

    result = np.empty(
        len(times),
        dtype=np.float32
    )

    for i in range(
        len(times)
    ):

        tt = times[i]
        ss = sensors[i]

        result[i] = np.std(
            speed_matrix[
                tt - window:
                tt,
                ss
            ]
        )

    return result


# ------------------------------------------------------------
# Rolling means
# ------------------------------------------------------------

for window in [
    3,
    6,
    12,
    24
]:

    X_train_new[:, col] = (
        rolling_mean_for_pairs(
            t,
            s,
            window
        )
    )

    col += 1


# ------------------------------------------------------------
# Rolling std
# ------------------------------------------------------------

for window in [
    6,
    12,
    24
]:

    X_train_new[:, col] = (
        rolling_std_for_pairs(
            t,
            s,
            window
        )
    )

    col += 1


# ============================================================
# 5. TRAFFIC DYNAMICS
# ============================================================

current = speed_matrix[t, s]

lag1 = speed_matrix[t - 1, s]
lag3 = speed_matrix[t - 3, s]
lag6 = speed_matrix[t - 6, s]


X_train_new[:, col] = (
    current - lag1
)
col += 1


X_train_new[:, col] = (
    current - lag3
)
col += 1


X_train_new[:, col] = (
    current - lag6
)
col += 1


# ============================================================
# 6. SPEED VS ROLLING MEAN
# ============================================================

mean6 = rolling_mean_for_pairs(
    t,
    s,
    6
)

mean12 = rolling_mean_for_pairs(
    t,
    s,
    12
)


X_train_new[:, col] = (
    current / (mean6 + 1e-6)
)

col += 1


X_train_new[:, col] = (
    current / (mean12 + 1e-6)
)

col += 1


# ============================================================
# 7. WEATHER
# ============================================================

X_train_new[:, col] = (
    temperature[t]
)
col += 1


X_train_new[:, col] = (
    humidity[t]
)
col += 1


X_train_new[:, col] = (
    rain[t]
)
col += 1


X_train_new[:, col] = (
    weather_code[t]
)
col += 1


X_train_new[:, col] = (
    temp_humidity[t]
)
col += 1


X_train_new[:, col] = (
    rain_flag[t]
)
col += 1


X_train_new[:, col] = (
    rain_intensity[t]
)
col += 1


X_train_new[:, col] = (
    bad_weather[t]
)
col += 1


# ============================================================
# 8. TIME
# ============================================================

time_features = [
    hour,
    minute,
    dow,
    month,
    is_weekend,
    is_holiday,
    morning_rush,
    evening_rush,
    rush_hour,
    hour_sin,
    hour_cos,
    dow_sin,
    dow_cos,
    month_sin,
    month_cos
]


for arr in time_features:

    X_train_new[:, col] = (
        arr[t]
    )

    col += 1


# ============================================================
# 9. EVENTS
# ============================================================

X_train_new[:, col] = (
    event_count_time[t]
)

col += 1


X_train_new[:, col] = (
    has_event_time[t]
)

col += 1


# ============================================================
# 10. TARGET
# ============================================================

y_train_new = (
    speed_matrix[
        t + 3,
        s
    ]
    .astype(
        np.float32
    )
)


# ============================================================
# CLEAN
# ============================================================

del current
del lag1
del lag3
del lag6
del mean6
del mean12
del t
del s

gc.collect()


# ============================================================
# FINAL CHECK
# ============================================================

print()
print('=' * 60)
print('NEW TRAIN DATA')
print('=' * 60)

print(
    'X_train:',
    X_train_new.shape
)

print(
    'y_train:',
    y_train_new.shape
)

print(
    'X RAM:',
    round(
        X_train_new.nbytes / 1024**2,
        2
    ),
    'MB'
)

print(
    'y RAM:',
    round(
        y_train_new.nbytes / 1024**2,
        2
    ),
    'MB'
)


# ============================================================
# CRITICAL SANITY CHECK
# ============================================================

print()
print(
    'Current speed / target correlation:'
)

print(
    np.corrcoef(
        X_train_new[:, 1],
        y_train_new
    )[0, 1]
)


print(
    'Naive current-speed R²:'
)

from sklearn.metrics import r2_score

print(
    r2_score(
        y_train_new,
        X_train_new[:, 1]
    )
)
```

```python
# ============================================================
# V7 - SPATIAL FEATURES
# STEP 1: PREPARE ADJACENCY MATRIX
# ============================================================

import numpy as np
import pandas as pd
import gc

print('=' * 60)
print('PREPARING SPATIAL GRAPH')
print('=' * 60)


# ------------------------------------------------------------
# Get adjacency from loaded pickle
# ------------------------------------------------------------

# adj_data should be:
# [sensor_ids, sensor_id_mapping, adjacency_matrix]

adj_sensor_ids = list(
    adj_data[0]
)

adj_matrix = np.asarray(
    adj_data[2],
    dtype=np.float32
)


print(
    'Adjacency shape:',
    adj_matrix.shape
)

print(
    'Sensor count:',
    len(sensor_columns)
)


# ------------------------------------------------------------
# Check sensors
# ------------------------------------------------------------

traffic_sensor_ids = [
    str(x)
    for x in sensor_columns
]

graph_sensor_ids = [
    str(x)
    for x in adj_sensor_ids
]


print(
    'Traffic sensors:',
    len(traffic_sensor_ids)
)

print(
    'Graph sensors:',
    len(graph_sensor_ids)
)


# ------------------------------------------------------------
# Make sure both contain exactly the same sensors
# ------------------------------------------------------------

if set(traffic_sensor_ids) != set(graph_sensor_ids):

    missing_from_graph = (
        set(traffic_sensor_ids)
        -
        set(graph_sensor_ids)
    )

    missing_from_traffic = (
        set(graph_sensor_ids)
        -
        set(traffic_sensor_ids)
    )

    raise ValueError(
        'Sensor mismatch!\n'
        f'Missing from graph: {missing_from_graph}\n'
        f'Missing from traffic: {missing_from_traffic}'
    )


# ------------------------------------------------------------
# Reorder adjacency to EXACT traffic sensor order
# ------------------------------------------------------------

graph_index = {
    sensor_id: i
    for i, sensor_id
    in enumerate(graph_sensor_ids)
}


reorder = np.array(
    [
        graph_index[sensor_id]
        for sensor_id
        in traffic_sensor_ids
    ],
    dtype=np.int32
)


adj_matrix = adj_matrix[
    reorder
][:, reorder]


print()
print(
    'Reordered adjacency:',
    adj_matrix.shape
)


# ------------------------------------------------------------
# Remove self-connections
# ------------------------------------------------------------

adj_no_self = (
    adj_matrix.copy()
)

np.fill_diagonal(
    adj_no_self,
    0
)


# ------------------------------------------------------------
# Weighted normalization
# ------------------------------------------------------------

row_sum = (
    adj_no_self.sum(
        axis=1,
        keepdims=True
    )
)


# Prevent division by zero
row_sum[
    row_sum == 0
] = 1.0


W = (
    adj_no_self
    /
    row_sum
).astype(
    np.float32
)


print()
print(
    'Graph ready!'
)

print(
    'Non-zero edges:',
    np.count_nonzero(
        adj_no_self
    )
)

print(
    'W min:',
    W.min()
)

print(
    'W max:',
    W.max()
)

print(
    'W row sums:',
    W.sum(
        axis=1
    )[:10]
)

print('=' * 60)
```

```python
# ============================================================
# V7 - SPATIAL FEATURES
# STEP 2: BUILD SPATIAL MATRICES
# ============================================================

print('=' * 60)
print('BUILDING SPATIAL FEATURES')
print('=' * 60)


# ------------------------------------------------------------
# Weighted neighbor mean
# ------------------------------------------------------------

neighbor_mean = (
    speed_matrix
    @
    W.T
).astype(
    np.float32
)


print(
    'neighbor_mean:',
    neighbor_mean.shape
)


# ------------------------------------------------------------
# Neighbor mean at lag 1
# ------------------------------------------------------------

neighbor_lag1 = np.zeros_like(
    neighbor_mean,
    dtype=np.float32
)

neighbor_lag1[1:] = (
    neighbor_mean[:-1]
)


# ------------------------------------------------------------
# Neighbor mean at lag 3
# ------------------------------------------------------------

neighbor_lag3 = np.zeros_like(
    neighbor_mean,
    dtype=np.float32
)

neighbor_lag3[3:] = (
    neighbor_mean[:-3]
)


# ------------------------------------------------------------
# Neighbor change
# ------------------------------------------------------------

neighbor_change = (
    neighbor_mean
    -
    neighbor_lag1
).astype(
    np.float32
)


# ------------------------------------------------------------
# Unweighted neighbor std
# ------------------------------------------------------------

neighbor_count = (
    (adj_no_self > 0)
    .sum(
        axis=1
    )
)


neighbor_sum = (
    speed_matrix
    @
    (adj_no_self > 0).T.astype(
        np.float32
    )
)


neighbor_sum_sq = (
    (speed_matrix ** 2)
    @
    (adj_no_self > 0).T.astype(
        np.float32
    )
)


neighbor_count_safe = (
    neighbor_count
    .copy()
    .astype(
        np.float32
    )
)

neighbor_count_safe[
    neighbor_count_safe == 0
] = 1


neighbor_unweighted_mean = (
    neighbor_sum
    /
    neighbor_count_safe
)


neighbor_var = (
    neighbor_sum_sq
    /
    neighbor_count_safe
    -
    neighbor_unweighted_mean ** 2
)


neighbor_var = np.maximum(
    neighbor_var,
    0
)


neighbor_std = np.sqrt(
    neighbor_var
).astype(
    np.float32
)


del neighbor_sum
del neighbor_sum_sq
del neighbor_unweighted_mean
del neighbor_var

gc.collect()


print(
    'neighbor_lag1:',
    neighbor_lag1.shape
)

print(
    'neighbor_lag3:',
    neighbor_lag3.shape
)

print(
    'neighbor_change:',
    neighbor_change.shape
)

print(
    'neighbor_std:',
    neighbor_std.shape
)

print('=' * 60)
```

```python
# ============================================================
# V7 - DIRECT FEATURE BUILDER
# Exact (time, sensor) pairs
# ============================================================

def build_direct_features(
    times,
    sensors,
    selected_features,
    include_spatial=True
):

    n = len(times)

    X = np.empty(
        (
            n,
            len(selected_features)
        ),
        dtype=np.float32
    )

    feature_to_col = {
        name: i
        for i, name in enumerate(
            selected_features
        )
    }


    # ========================================================
    # BASIC TRAFFIC FEATURES
    # ========================================================

    if 'sensor_id_encoded' in feature_to_col:

        X[:, feature_to_col[
            'sensor_id_encoded'
        ]] = sensor_encoded[
            sensors
        ]


    if 'speed' in feature_to_col:

        X[:, feature_to_col[
            'speed'
        ]] = speed_matrix[
            times,
            sensors
        ]


    # ========================================================
    # LAGS
    # ========================================================

    lag_map = {
        'lag_1': 1,
        'lag_2': 2,
        'lag_3': 3,
        'lag_4': 4,
        'lag_6': 6,
        'lag_9': 9,
        'lag_12': 12,
        'lag_18': 18,
        'lag_24': 24
    }


    for feature, lag in lag_map.items():

        if feature in feature_to_col:

            X[:, feature_to_col[
                feature
            ]] = speed_matrix[
                times - lag,
                sensors
            ]


    # ========================================================
    # ROLLING
    # ========================================================

    rolling_cache = {}


    for window in [
        3,
        6,
        12,
        24
    ]:

        needed = (
            f'rolling_mean_{window}'
            in feature_to_col
        )

        if needed:

            values = np.empty(
                n,
                dtype=np.float32
            )

            for i in range(n):

                tt = times[i]
                ss = sensors[i]

                values[i] = np.mean(
                    speed_matrix[
                        tt - window:tt,
                        ss
                    ]
                )

            rolling_cache[
                f'mean_{window}'
            ] = values


            X[:, feature_to_col[
                f'rolling_mean_{window}'
            ]] = values


    # ========================================================
    # ROLLING STD
    # ========================================================

    for window in [
        6,
        12,
        24
    ]:

        feature = (
            f'rolling_std_{window}'
        )

        if feature in feature_to_col:

            values = np.empty(
                n,
                dtype=np.float32
            )

            for i in range(n):

                tt = times[i]
                ss = sensors[i]

                values[i] = np.std(
                    speed_matrix[
                        tt - window:tt,
                        ss
                    ]
                )

            X[:, feature_to_col[
                feature
            ]] = values


    # ========================================================
    # SPEED CHANGES
    # ========================================================

    change_map = {
        'speed_change_1': 1,
        'speed_change_3': 3,
        'speed_change_6': 6
    }


    current = speed_matrix[
        times,
        sensors
    ]


    for feature, lag in change_map.items():

        if feature in feature_to_col:

            X[:, feature_to_col[
                feature
            ]] = (
                current
                -
                speed_matrix[
                    times - lag,
                    sensors
                ]
            )


    # ========================================================
    # SPEED VS MEAN
    # ========================================================

    for window in [
        6,
        12
    ]:

        feature = (
            f'speed_vs_mean_{window}'
        )

        if feature in feature_to_col:

            if (
                f'mean_{window}'
                in rolling_cache
            ):

                mean_value = rolling_cache[
                    f'mean_{window}'
                ]

            else:

                mean_value = np.empty(
                    n,
                    dtype=np.float32
                )

                for i in range(n):

                    tt = times[i]
                    ss = sensors[i]

                    mean_value[i] = np.mean(
                        speed_matrix[
                            tt-window:tt,
                            ss
                        ]
                    )

            X[:, feature_to_col[
                feature
            ]] = (
                current
                /
                (
                    mean_value
                    + 1e-6
                )
            )


    # ========================================================
    # WEATHER
    # ========================================================

    weather_map = {
        'temperature_2m': temperature,
        'relative_humidity_2m': humidity,
        'rain': rain,
        'weather_code': weather_code,
        'temp_humidity_interaction':
            temp_humidity,
        'rain_flag': rain_flag,
        'rain_intensity': rain_intensity,
        'bad_weather': bad_weather
    }


    for feature, values in weather_map.items():

        if feature in feature_to_col:

            X[:, feature_to_col[
                feature
            ]] = values[
                times
            ]


    # ========================================================
    # TIME
    # ========================================================

    time_map = {
        'hour': hour,
        'minute': minute,
        'day_of_week': dow,
        'month': month,

        'is_weekend': is_weekend,
        'is_holiday': is_holiday,

        'is_morning_rush':
            morning_rush,

        'is_evening_rush':
            evening_rush,

        'is_rush_hour':
            rush_hour,

        'hour_sin': hour_sin,
        'hour_cos': hour_cos,

        'dow_sin': dow_sin,
        'dow_cos': dow_cos,

        'month_sin': month_sin,
        'month_cos': month_cos
    }


    for feature, values in time_map.items():

        if feature in feature_to_col:

            X[:, feature_to_col[
                feature
            ]] = values[
                times
            ]


    # ========================================================
    # EVENTS
    # ========================================================

    if 'event_count' in feature_to_col:

        X[:, feature_to_col[
            'event_count'
        ]] = event_count_time[
            times
        ]


    if 'has_event' in feature_to_col:

        X[:, feature_to_col[
            'has_event'
        ]] = has_event_time[
            times
        ]


    # ========================================================
    # SPATIAL
    # ========================================================

    if include_spatial:

        spatial_map = {

            'neighbor_mean':
                neighbor_mean,

            'neighbor_lag1':
                neighbor_lag1,

            'neighbor_lag3':
                neighbor_lag3,

            'neighbor_change':
                neighbor_change,

            'neighbor_std':
                neighbor_std
        }


        for feature, values in spatial_map.items():

            if feature in feature_to_col:

                X[:, feature_to_col[
                    feature
                ]] = values[
                    times,
                    sensors
                ]


    return X
```

```python
# ==========================================
# 4. Sensor Locations
# ==========================================

sensor_locations = pd.read_csv(
    '../../dataset/METR-LA/sensor_locations.csv'
)

sensor_locations['sensor_id'] = (
    sensor_locations['sensor_id']
    .astype(str)
)

sensor_locations = (
    sensor_locations
    .set_index('sensor_id')
    .loc[sensor_columns]
    .reset_index()
)


coords = sensor_locations[
    [
        'latitude',
        'longitude'
    ]
].to_numpy(
    dtype=np.float32
)


print(
    'Coordinates:',
    coords.shape
)


# ==========================================
# 5. Nearest Neighbors
# ==========================================

K = 5

knn = NearestNeighbors(
    n_neighbors=K + 1
)

knn.fit(coords)

distances, indices = (
    knn.kneighbors(coords)
)


# Remove self
neighbor_indices = indices[:, 1:]

print(
    'Neighbor matrix:',
    neighbor_indices.shape
)
```

```python
# ==========================================
# 6. Spatial Features
# ==========================================

neighbor_speed = (
    speed_matrix[
        :,
        neighbor_indices
    ]
)


print(
    'Neighbor speed shape:',
    neighbor_speed.shape
)
```

```python
neighbor_mean = (
    np.mean(
        neighbor_speed,
        axis=2
    )
    .astype(np.float32)
)

neighbor_std = (
    np.std(
        neighbor_speed,
        axis=2
    )
    .astype(np.float32)
)

neighbor_min = (
    np.min(
        neighbor_speed,
        axis=2
    )
    .astype(np.float32)
)

neighbor_max = (
    np.max(
        neighbor_speed,
        axis=2
    )
    .astype(np.float32)
)
```

```python
neighbor_diff = (
    speed_matrix
    - neighbor_mean
).astype(
    np.float32
)
```

```python
neighbor_ratio = (
    speed_matrix
    / np.maximum(
        neighbor_mean,
        1.0
    )
).astype(
    np.float32
)
```

```python
# ==========================================
# 7. Temporal Features
# ==========================================

lag_1 = np.roll(
    speed_matrix,
    1,
    axis=0
)

lag_3 = np.roll(
    speed_matrix,
    3,
    axis=0
)

lag_6 = np.roll(
    speed_matrix,
    6,
    axis=0
)

lag_12 = np.roll(
    speed_matrix,
    12,
    axis=0
)

lag_18 = np.roll(
    speed_matrix,
    18,
    axis=0
)
```

```python
rolling_mean_3 = (
    pd.DataFrame(speed_matrix)
    .shift(1)
    .rolling(3)
    .mean()
    .to_numpy(
        dtype=np.float32
    )
)

rolling_mean_6 = (
    pd.DataFrame(speed_matrix)
    .shift(1)
    .rolling(6)
    .mean()
    .to_numpy(
        dtype=np.float32
    )
)

rolling_mean_12 = (
    pd.DataFrame(speed_matrix)
    .shift(1)
    .rolling(12)
    .mean()
    .to_numpy(
        dtype=np.float32
    )
)

rolling_std_6 = (
    pd.DataFrame(speed_matrix)
    .shift(1)
    .rolling(6)
    .std()
    .to_numpy(
        dtype=np.float32
    )
)
```

```python
# ==========================================
# 9. Time Features
# ==========================================

time = base_df['time']

hour = (
    time.dt.hour
    .to_numpy(
        dtype=np.float32
    )
)

minute = (
    time.dt.minute
    .to_numpy(
        dtype=np.float32
    )
)

day_of_week = (
    time.dt.dayofweek
    .to_numpy(
        dtype=np.float32
    )
)

month = (
    time.dt.month
    .to_numpy(
        dtype=np.float32
    )
)

is_weekend = (
    day_of_week >= 5
).astype(
    np.float32
)

is_holiday = (
    base_df['Holiday']
    .ne('None')
    .to_numpy(
        dtype=np.float32
    )
)


hour_sin = np.sin(
    2 * np.pi * hour / 24
).astype(
    np.float32
)

hour_cos = np.cos(
    2 * np.pi * hour / 24
).astype(
    np.float32
)

dow_sin = np.sin(
    2 * np.pi * day_of_week / 7
).astype(
    np.float32
)

dow_cos = np.cos(
    2 * np.pi * day_of_week / 7
).astype(
    np.float32
)
```

```python
# ==========================================
# V3 Dataset Builder - Clean
# ==========================================

def build_v3_dataset(
    time_indices,
    speed_matrix,
    lag_1,
    lag_3,
    lag_6,
    lag_12,
    lag_18,
    rolling_mean_3,
    rolling_mean_6,
    rolling_mean_12,
    rolling_std_6,
    neighbor_mean,
    neighbor_std,
    neighbor_min,
    neighbor_max,
    neighbor_diff,
    neighbor_ratio,
    temperature,
    humidity,
    rain,
    weather_code,
    hour,
    minute,
    day_of_week,
    month,
    is_weekend,
    is_holiday,
    hour_sin,
    hour_cos,
    dow_sin,
    dow_cos,
    sensor_encoded,
    target=None
):

    n_times = len(time_indices)
    n_sensors = speed_matrix.shape[1]

    n_features = 31

    X = np.empty(
        (
            n_times * n_sensors,
            n_features
        ),
        dtype=np.float32
    )

    row = 0

    for t in time_indices:

        end_row = row + n_sensors

        X[row:end_row, 0] = sensor_encoded
        X[row:end_row, 1] = speed_matrix[t]

        X[row:end_row, 2] = lag_1[t]
        X[row:end_row, 3] = lag_3[t]
        X[row:end_row, 4] = lag_6[t]
        X[row:end_row, 5] = lag_12[t]
        X[row:end_row, 6] = lag_18[t]

        X[row:end_row, 7] = rolling_mean_3[t]
        X[row:end_row, 8] = rolling_mean_6[t]
        X[row:end_row, 9] = rolling_mean_12[t]
        X[row:end_row, 10] = rolling_std_6[t]

        X[row:end_row, 11] = neighbor_mean[t]
        X[row:end_row, 12] = neighbor_std[t]
        X[row:end_row, 13] = neighbor_min[t]
        X[row:end_row, 14] = neighbor_max[t]
        X[row:end_row, 15] = neighbor_diff[t]
        X[row:end_row, 16] = neighbor_ratio[t]

        X[row:end_row, 17] = temperature[t]
        X[row:end_row, 18] = humidity[t]
        X[row:end_row, 19] = rain[t]
        X[row:end_row, 20] = weather_code[t]

        X[row:end_row, 21] = hour[t]
        X[row:end_row, 22] = minute[t]
        X[row:end_row, 23] = day_of_week[t]
        X[row:end_row, 24] = month[t]
        X[row:end_row, 25] = is_weekend[t]
        X[row:end_row, 26] = is_holiday[t]

        X[row:end_row, 27] = hour_sin[t]
        X[row:end_row, 28] = hour_cos[t]
        X[row:end_row, 29] = dow_sin[t]
        X[row:end_row, 30] = dow_cos[t]

        row = end_row


    if target is not None:

        y = np.empty(
            n_times * n_sensors,
            dtype=np.float32
        )

        row = 0

        for t in time_indices:

            end_row = row + n_sensors

            y[row:end_row] = target[t]

            row = end_row

        return X, y


    return X
```

```python
# ==========================================
# Build Selected Training Samples
# ==========================================

def build_selected_v3_dataset(
    time_positions,
    sensor_positions,
    speed_matrix,
    lag_1,
    lag_3,
    lag_6,
    lag_12,
    lag_18,
    rolling_mean_3,
    rolling_mean_6,
    rolling_mean_12,
    rolling_std_6,
    neighbor_mean,
    neighbor_std,
    neighbor_min,
    neighbor_max,
    neighbor_diff,
    neighbor_ratio,
    temperature,
    humidity,
    rain,
    weather_code,
    hour,
    minute,
    day_of_week,
    month,
    is_weekend,
    is_holiday,
    hour_sin,
    hour_cos,
    dow_sin,
    dow_cos,
    sensor_encoded,
    target
):

    n = len(time_positions)

    X = np.empty(
        (n, 31),
        dtype=np.float32
    )

    X[:, 0] = sensor_encoded[
        sensor_positions
    ]

    X[:, 1] = speed_matrix[
        time_positions,
        sensor_positions
    ]

    X[:, 2] = lag_1[
        time_positions,
        sensor_positions
    ]

    X[:, 3] = lag_3[
        time_positions,
        sensor_positions
    ]

    X[:, 4] = lag_6[
        time_positions,
        sensor_positions
    ]

    X[:, 5] = lag_12[
        time_positions,
        sensor_positions
    ]

    X[:, 6] = lag_18[
        time_positions,
        sensor_positions
    ]

    X[:, 7] = rolling_mean_3[
        time_positions,
        sensor_positions
    ]

    X[:, 8] = rolling_mean_6[
        time_positions,
        sensor_positions
    ]

    X[:, 9] = rolling_mean_12[
        time_positions,
        sensor_positions
    ]

    X[:, 10] = rolling_std_6[
        time_positions,
        sensor_positions
    ]

    X[:, 11] = neighbor_mean[
        time_positions,
        sensor_positions
    ]

    X[:, 12] = neighbor_std[
        time_positions,
        sensor_positions
    ]

    X[:, 13] = neighbor_min[
        time_positions,
        sensor_positions
    ]

    X[:, 14] = neighbor_max[
        time_positions,
        sensor_positions
    ]

    X[:, 15] = neighbor_diff[
        time_positions,
        sensor_positions
    ]

    X[:, 16] = neighbor_ratio[
        time_positions,
        sensor_positions
    ]

    X[:, 17] = temperature[
        time_positions
    ]

    X[:, 18] = humidity[
        time_positions
    ]

    X[:, 19] = rain[
        time_positions
    ]

    X[:, 20] = weather_code[
        time_positions
    ]

    X[:, 21] = hour[
        time_positions
    ]

    X[:, 22] = minute[
        time_positions
    ]

    X[:, 23] = day_of_week[
        time_positions
    ]

    X[:, 24] = month[
        time_positions
    ]

    X[:, 25] = is_weekend[
        time_positions
    ]

    X[:, 26] = is_holiday[
        time_positions
    ]

    X[:, 27] = hour_sin[
        time_positions
    ]

    X[:, 28] = hour_cos[
        time_positions
    ]

    X[:, 29] = dow_sin[
        time_positions
    ]

    X[:, 30] = dow_cos[
        time_positions
    ]

    y = target[
        time_positions,
        sensor_positions
    ].astype(
        np.float32
    )

    return X, y
```

```python
# ==========================================
# Traffic Sequence Dataset
# ==========================================

class TrafficSequenceDataset(Dataset):

    def __init__(
        self,
        speed_matrix,
        start_index,
        end_index,
        history=12,
        horizon=3,
        mean=0.0,
        std=1.0
    ):

        self.speed_matrix = speed_matrix

        self.history = history
        self.horizon = horizon

        self.mean = np.float32(mean)
        self.std = np.float32(std)

        self.indices = np.arange(
            start_index + history,
            end_index - horizon + 1,
            dtype=np.int32
        )


    def __len__(self):

        return len(
            self.indices
        )


    def __getitem__(
        self,
        idx
    ):

        t = int(
            self.indices[idx]
        )


        # ------------------------------
        # Past
        # ------------------------------

        x = self.speed_matrix[
            t - self.history:t
        ]


        # ------------------------------
        # Future
        # ------------------------------

        y = self.speed_matrix[
            t:t + self.horizon
        ]


        # ------------------------------
        # Normalize
        # ------------------------------

        x = (
            x - self.mean
        ) / self.std

        y = (
            y - self.mean
        ) / self.std


        # ------------------------------
        # Add feature dimension
        # ------------------------------

        x = x[
            :, :, np.newaxis
        ]


        # ------------------------------
        # NumPy → Torch
        # ------------------------------

        x = torch.from_numpy(
            x.astype(
                np.float32
            )
        )

        y = torch.from_numpy(
            y.astype(
                np.float32
            )
        )


        return x, y
```

```python
# ==========================================
# Normalize Graph
# ==========================================

A = torch.tensor(
    adj_aligned,
    dtype=torch.float32
)


degree = A.sum(
    dim=1
)


degree_inv_sqrt = torch.pow(
    degree,
    -0.5
)

degree_inv_sqrt[
    torch.isinf(
        degree_inv_sqrt
    )
] = 0.0


D_inv_sqrt = torch.diag(
    degree_inv_sqrt
)


A_norm = (
    D_inv_sqrt
    @ A
    @ D_inv_sqrt
)


A_norm = A_norm.to(
    device
)


print(
    'A shape:',
    A_norm.shape
)
```

```python
# ==========================================
# Graph Convolution
# ==========================================

class GraphConv(nn.Module):

    def __init__(
        self,
        in_channels,
        out_channels
    ):

        super().__init__()

        self.linear = nn.Linear(
            in_channels,
            out_channels
        )


    def forward(
        self,
        x,
        A
    ):

        # x:
        # [batch, time, nodes, features]

        x = torch.einsum(
            'ij,btjf->btif',
            A,
            x
        )

        x = self.linear(
            x
        )

        return x
```

```python
# ==========================================
# Temporal Block
# ==========================================

class TemporalBlock(nn.Module):

    def __init__(
        self,
        in_channels,
        out_channels,
        kernel_size=3
    ):

        super().__init__()

        self.conv = nn.Conv2d(
            in_channels,
            out_channels,
            kernel_size=(
                kernel_size,
                1
            ),
            padding=(
                kernel_size // 2,
                0
            )
        )

        self.bn = nn.BatchNorm2d(
            out_channels
        )


    def forward(
        self,
        x
    ):

        # [B, T, N, C]
        # →
        # [B, C, T, N]

        x = x.permute(
            0,
            3,
            1,
            2
        )

        x = self.conv(
            x
        )

        x = self.bn(
            x
        )

        x = F.relu(
            x
        )

        # →
        # [B, T, N, C]

        x = x.permute(
            0,
            2,
            3,
            1
        )

        return x
```

```python
# ==========================================
# STGCN
# ==========================================

class STGCN(nn.Module):

    def __init__(
        self,
        num_nodes,
        input_features=1,
        hidden_channels=32,
        horizon=3
    ):

        super().__init__()

        self.num_nodes = num_nodes

        self.horizon = horizon


        self.temporal1 = TemporalBlock(
            input_features,
            hidden_channels
        )


        self.graph1 = GraphConv(
            hidden_channels,
            hidden_channels
        )


        self.temporal2 = TemporalBlock(
            hidden_channels,
            hidden_channels
        )


        self.graph2 = GraphConv(
            hidden_channels,
            hidden_channels
        )


        self.output = nn.Linear(
            hidden_channels,
            horizon
        )


    def forward(
        self,
        x,
        A
    ):

        # ----------------------------------
        # Temporal
        # ----------------------------------

        x = self.temporal1(
            x
        )


        # ----------------------------------
        # Spatial
        # ----------------------------------

        x = self.graph1(
            x,
            A
        )

        x = F.relu(
            x
        )


        # ----------------------------------
        # Temporal
        # ----------------------------------

        x = self.temporal2(
            x
        )


        # ----------------------------------
        # Spatial
        # ----------------------------------

        x = self.graph2(
            x,
            A
        )

        x = F.relu(
            x
        )


        # ----------------------------------
        # Last time step
        # ----------------------------------

        x = x[:, -1]

        # [B, N, hidden]


        # ----------------------------------
        # Predict horizon
        # ----------------------------------

        x = self.output(
            x
        )

        # [B, N, horizon]


        x = x.permute(
            0,
            2,
            1
        )

        # [B, horizon, N]


        return x
```

```python
# ==========================================
# Training Configuration
# ==========================================

import copy
import numpy as np
import torch
import torch.nn as nn


learning_rate = 0.001
weight_decay = 1e-4

epochs = 20

patience = 4

gradient_clip = 1.0


optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=learning_rate,
    weight_decay=weight_decay
)


criterion = nn.HuberLoss(
    delta=1.0
)


scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode='min',
    factor=0.5,
    patience=2
)


print('Optimizer:', optimizer)
print('Loss:', criterion)
```

```python
# ==========================================
# Validation Split
# ==========================================

train_start = start

train_stop = train_end + 1

validation_start = int(
    train_start
    + 0.90 * (
        train_stop - train_start
    )
)


validation_dataset = TrafficSequenceDataset(
    speed_matrix=speed_matrix,

    start_index=validation_start,

    end_index=train_stop,

    history=HISTORY,

    horizon=HORIZON,

    mean=train_mean,

    std=train_std
)


validation_loader = DataLoader(
    validation_dataset,

    batch_size=BATCH_SIZE,

    shuffle=False,

    num_workers=0,

    pin_memory=False
)


print(
    'Validation sequences:',
    len(validation_dataset)
)

print(
    'Validation batches:',
    len(validation_loader)
)
```

```python
# ==========================================
# Training Loop
# ==========================================

best_val_loss = float('inf')

best_model_state = None

epochs_without_improvement = 0


use_amp = (
    device.type == 'cuda'
)


if use_amp:

    scaler = torch.amp.GradScaler(
        'cuda'
    )


for epoch in range(
    1,
    epochs + 1
):

    # ======================================
    # TRAIN
    # ======================================

    model.train()

    train_loss_sum = 0.0

    train_batches = 0


    for batch_idx, (
        X,
        Y
    ) in enumerate(
        train_loader
    ):

        X = X.to(
            device,
            non_blocking=True
        )

        Y = Y.to(
            device,
            non_blocking=True
        )


        optimizer.zero_grad(
            set_to_none=True
        )


        if use_amp:

            with torch.autocast(
                device_type='cuda',
                dtype=torch.float16
            ):

                prediction = model(
                    X,
                    A_norm
                )

                loss = criterion(
                    prediction,
                    Y
                )


            scaler.scale(
                loss
            ).backward()


            scaler.unscale_(
                optimizer
            )


            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                gradient_clip
            )


            scaler.step(
                optimizer
            )

            scaler.update()


        else:

            prediction = model(
                X,
                A_norm
            )

            loss = criterion(
                prediction,
                Y
            )


            loss.backward()


            torch.nn.utils.clip_grad_norm_(
                model.parameters(),
                gradient_clip
            )


            optimizer.step()


        train_loss_sum += (
            loss.item()
        )

        train_batches += 1


    train_loss = (
        train_loss_sum
        /
        train_batches
    )


    # ======================================
    # VALIDATION
    # ======================================

    model.eval()

    val_loss_sum = 0.0

    val_batches = 0


    with torch.no_grad():

        for X, Y in validation_loader:

            X = X.to(
                device,
                non_blocking=True
            )

            Y = Y.to(
                device,
                non_blocking=True
            )


            if use_amp:

                with torch.autocast(
                    device_type='cuda',
                    dtype=torch.float16
                ):

                    prediction = model(
                        X,
                        A_norm
                    )

                    val_loss = criterion(
                        prediction,
                        Y
                    )

            else:

                prediction = model(
                    X,
                    A_norm
                )

                val_loss = criterion(
                    prediction,
                    Y
                )


            val_loss_sum += (
                val_loss.item()
            )

            val_batches += 1


    val_loss = (
        val_loss_sum
        /
        val_batches
    )


    # ======================================
    # Scheduler
    # ======================================

    scheduler.step(
        val_loss
    )


    current_lr = (
        optimizer.param_groups[0]['lr']
    )


    # ======================================
    # Save Best
    # ======================================

    if val_loss < best_val_loss:

        best_val_loss = val_loss

        best_model_state = copy.deepcopy(
            model.state_dict()
        )

        epochs_without_improvement = 0

        print(
            f'Epoch {epoch:02d} | '
            f'Train: {train_loss:.5f} | '
            f'Val: {val_loss:.5f} | '
            f'LR: {current_lr:.6f} | '
            f'★ BEST'
        )

    else:

        epochs_without_improvement += 1

        print(
            f'Epoch {epoch:02d} | '
            f'Train: {train_loss:.5f} | '
            f'Val: {val_loss:.5f} | '
            f'LR: {current_lr:.6f}'
        )


    # ======================================
    # Early Stopping
    # ======================================

    if (
        epochs_without_improvement
        >= patience
    ):

        print(
            'Early stopping!'
        )

        break


# ==========================================
# Restore Best Model
# ==========================================

if best_model_state is not None:

    model.load_state_dict(
        best_model_state
    )


print(
    'Training complete!'
)

print(
    'Best validation loss:',
    best_val_loss
)
```
